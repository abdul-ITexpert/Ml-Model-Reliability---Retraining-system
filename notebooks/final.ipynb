```python
import pandas as pd
import os

production_labels = []

for i in range(1, 6):

    label_path = fr"data\production\labels_{i:03d}.csv"

    if os.path.exists(label_path):

        y_batch = pd.read_csv(label_path)

        # Convert DataFrame to Series
        y_batch = y_batch.iloc[:, 0]

        production_labels.append(y_batch)

        print(
            f"Batch {i} labels:",
            len(y_batch)
        )

    else:

        print(
            f"⚠️ File not found: {label_path}"
        )
```

```python
print(
    "Total predictions:",
    len(logs)
)
```

```python
print(
    logs["model_version"].value_counts()
)
```

```python
print(
    logs["prediction"].value_counts()
)
```

```python
average_probability = logs[
    "prediction_probability"
].mean()

print(
    "Average probability:",
    average_probability
)
```

```python
logs["risk_level"] = pd.cut(
    logs["prediction_probability"],
    bins=[0, 0.3, 0.7, 1],
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)
```

```python
print(
    logs["risk_level"].value_counts()
)
```

```python
production_labels = pd.concat(
    production_labels,
    ignore_index=True
)

print(
    "Total production labels:",
    len(production_labels)
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
y_true = production_labels

y_pred = logs["prediction"]

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred
)

recall = recall_score(
    y_true,
    y_pred
)

f1 = f1_score(
    y_true,
    y_pred
)

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1:", f1)
```

```python
import pandas as pd
import os

production_data = []

for i in range(1, 6):

    data_path = fr"data\production\batch_{i:03d}.csv"

    if os.path.exists(data_path):

        X_batch = pd.read_csv(data_path)

        production_data.append(X_batch)

        print(
            f"Batch {i} data:",
            X_batch.shape
        )

    else:

        print(
            f"⚠️ File not found: {data_path}"
        )


# Combine all production batches
production_data = pd.concat(
    production_data,
    ignore_index=True
)


# Save combined production data
output_path = r"data\processed\production_data.csv"

production_data.to_csv(
    output_path,
    index=False
)


print("\nProduction data created successfully!")
print("Shape:", production_data.shape)
print("Saved at:", output_path)
```
